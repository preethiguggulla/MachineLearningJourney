# Day17_RandomForest_Hyperparameters


```python
import pandas as pd
```

```python
df=pd.read_csv("/content/train.csv")
```

**Cleaning**

```python
df=df.drop(["PassengerId","Name","Ticket"],axis=1)
df.info()
```

```python
df["Age"]=df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"]=df["Embarked"].fillna(df["Embarked"].mode())
```

```python
df["Sex"]=df["Sex"].map({"male":0,"female":1})
```

```python
df["Sex"].value_counts(dropna=False)
```

```python
df=pd.get_dummies(df,columns=["Embarked"],drop_first=True)
```

```python
df.columns
```

```python
df.isnull().sum()
df.dtypes
```

```python
#Logistic Regression
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
y=df["Survived"]
df=df.drop(["Survived"],axis=1)
x=df
```

```python
x.info()
```

```python
x = x.drop("Cabin",axis=1)
from sklearn.linear_model import LogisticRegression
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
lr=LogisticRegression(max_iter=1000)
lr.fit(x_train,y_train)
train_acc_lr=accuracy_score(y_train,lr.predict(x_train))
test_acc_lr=accuracy_score(y_test,lr.predict(x_test))
train_acc_lr,test_acc_lr
```

```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier(max_depth=3,random_state=42)
dt.fit(x_train,y_train)
train_acc_dt=accuracy_score(y_train,dt.predict(x_train))
test_acc_dt=accuracy_score(y_test,dt.predict(x_test))
train_acc_dt,test_acc_dt
```

```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier(random_state=42)
rf.fit(x_train,y_train)
train_acc_rf=accuracy_score(y_train,rf.predict(x_train))
test_acc_rf=accuracy_score(y_test,rf.predict(x_test))
train_acc_rf,test_acc_rf
```

# Day18_ModelTuning

max_depth → controls complexity (overfitting)

n_estimators → number of trees (stability)

min_samples_split -> defines the minimum number of samples required to split an internal node. It acts as a regularization parameter to control the complexity of individual trees within the forest and prevent overfitting.

min_samples_leaf -> the minimum number of samples that must be present in a leaf node (terminal node) after a split. Its primary purpose is to control the complexity of individual decision trees within the forest and prevent overfitting.

## Day 20-how to evaluate a model beyond accuracy and how confident the model

```python
y_prob = lr.predict_proba(x_test)
y_prob[:5]
```

What do these two columns represent?
The first column represents the probability of class 0 (Not Survived) and the second column represents the probability of class 1 (Survived) for each test sample.

```python
from sklearn.metrics import roc_auc_score
roc_auc=roc_auc_score(y_test,y_prob[:,1])
roc_auc
```

Is ROC-AUC better or worse than accuracy here?and why?
ROC–AUC and accuracy are NOT comparable numbers

ROC–AUC measures ranking quality

Accuracy depends on chosen threshold (default = 0.5)

ROC–AUC is not better or worse than accuracy; it measures how well the model separates classes across all thresholds, while accuracy depends on a fixed threshold.

```python
import numpy as np
y_pred_custom=(y_prob[:,1]>0.3).astype(int)
y_pred_custom
```

```python
y_pred_custom=(y_prob[:,1]>0.5).astype(int)
y_pred_custom
```

Lower threshold (0.3) → predicts more survivors

Higher threshold (0.5) → stricter, fewer positives

You don’t need accuracy here, just understanding behavior.
What changed?
there are few differences when thresholds changes

Which errors increased/decreased?
Changing the threshold changes how confident the model must be before predicting survival, which affects false positives and false negatives.

# Day 21: Precision–Recall & Threshold Trade-offs

**Task 1 Precision & Recall intuition **

Precision = Of all predicted survived, how many actually survived?

Recall = Of all people who survived, how many did the model catch?

If we want to avoid false positives, do we increase precision or recall?
Precision
If we want to avoid missing true survivors, do we increase precision or recall?
Recall


```python
y_pred=lr.predict(x_test)
```

```python
from sklearn.metrics import precision_score,recall_score
precision=precision_score(y_test,y_pred)
recall=recall_score(y_test,y_pred)
precision,recall
```

Is your model more precision-oriented or recall-oriented?
Precision

```python
import numpy as np
y_pred_03=(y_prob[:,1]>0.3).astype(int)
y_pred_05=(y_prob[:,1]>0.5).astype(int)
y_pred_07=(y_prob[:,1]>0.7).astype(int)
```

```python
for t,yp in [(0.3,y_pred_03),(0.5,y_pred_05),(0.7,y_pred_07)]:
  print(t,precision_score(y_test,yp),recall_score(y_test,yp))
```

When threshold ↓, what happens to recall?
recall increaces

When threshold ↑, what happens to precision?
precision increases

Why is there no single best threshold?
it changes based on the dataset and models

There is no single best threshold because the optimal trade-off between precision and recall depends on the problem’s cost of false positives vs false negatives, the dataset distribution, and the business objective.

# Day 22 — ROC Curve & Threshold Visualization

```python
from sklearn.metrics import roc_curve
fpr,tpr,thresholds=roc_curve(y_test,y_prob[:,1])
fpr[:5],tpr[:5],thresholds[:5]
```

FPR = False Positive Rate
Out of all ACTUAL negatives, how many were wrongly predicted as positive
TPR = Recall (True Positive Rate)
out of all positives predicted by model how many are actually true

```python
import matplotlib.pyplot as plt
plt.plot(fpr,tpr)
plt.plot([0,1],[0,1],linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.show()
```

What does the diagonal line mean?
→ Random guessing
when model predicts only half of the positives are true positives


When is a model better?
→ Curve closer to top-left
curve is parallel to y axis

When threshold ↓, what happens to TPR?
increases

When threshold ↓, what happens to FPR?
increases
Lower threshold = model predicts more positives, so:

catches more true positives

but also creates more false positives

Why ROC curve does not depend on a single threshold?
ROC curve is built by sweeping across all possible thresholds, not one fixed threshold.

Why ROC-AUC is more reliable than accuracy for imbalanced data?
in real world data is not very clear and balanced , so when we relay only on accuracy model will not work well for real time data
